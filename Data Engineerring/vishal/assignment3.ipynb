```python
#question 6
def factorial(n):
    if n < 0:
        return 
    else:
        num = 1
        for i in range(1, n + 1):
            num = num * i
        return num
num1 = int(input("Enter a number: "))
result = factorial(num1)
print(result)
```

```python
#question7
def prime(n):
    if n <= 1:
        print("Not a prime number")
    else:
        num = 0
        for i in range(2, n):
            if n % i == 0:
                num = 1
                break
        if num == 0:
            print("Prime number")
        else:
            print("Not a prime number")
num2 = int(input("Enter a number: "))
prime(num2)
```

```python
#question 11
def uppercase(text):
    result = ""
    for char in text:
        if 'a' <= char <= 'z':
            result += chr(ord(char) - 32)
        else:
            result += char   
    print(f"Uppercase version: {result}")
num1 = input("Enter a string: ")
uppercase(num1)
```

```python
#question 16
def factors(n):
    m = []
    for i in range(1, n + 1):
        if n % i == 0:
            m.append(i)
    return m
num = int(input("Enter a number:"))
print(factors(num))
```

```python
#question 19..
def fibonacci(n):
    a=0
    b=1
    while a <= n:
        print(a, end=" ")
        a, b = b, a + b
    print()
num1=int(input("enter the value is: "))
fibonacci(num1)

def incr(n):
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            print(j, end=" ")
        print() 

incr(4)
```
