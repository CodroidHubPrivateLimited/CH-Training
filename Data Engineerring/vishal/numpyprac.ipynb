```python
#How can you detect and remove infinite values (inf, -inf) from a NumPy array?
arr=np.array([1,2,np.inf,-np.inf])
print(arr)
inf_indices=np.isinf(arr)
print(inf_indices)
arr[inf_indices]=0
print(arr)
```

```python
#How do you sort a NumPy array while keeping track of original indices
arr=np.array([3,1,4,2])
print(arr)
sorted_indices=np.argsort(arr)
print(sorted_indices)
sorted_arr=arr[sorted_indices]
print(sorted_arr)
```

```python
#How do you check if an array contains any null or invalid values?
arr=np.array([1,2,np.nan,4])
print(np.isnan(arr).any())
```

```python
#How can boolean masking be used to clean unwanted values from a NumPy array?
arr=np.array([1,2,-99,4,5,-99])
print(arr)
mask=arr!=-99
print(mask)
```
