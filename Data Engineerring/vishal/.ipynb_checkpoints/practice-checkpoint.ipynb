```python
total_sum = 0
for i in range(1, 51):
    total_sum += i
print("The total sum is:", total_sum)
```

```python
word = "sachdeva"
for char in word:
    print(char)
```

```python
text = "vishal"
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1

print("Number of vowels:", count)
```

```python
n = 4
factorial = 1

for i in range(1, n + 1):
    factorial *= i

print(f"The factorial of {n} is {factorial}")
```

```python
numbers = [12, 45, 2, 89, 34, 10]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("The largest number is:", largest)
```

```python
rows = 3
for i in range(1, rows + 1):
    print("*" * i)
```
