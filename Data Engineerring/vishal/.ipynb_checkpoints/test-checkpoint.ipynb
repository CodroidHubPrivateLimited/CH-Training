```python
name = "Vishal"
upper = 0
lower = 0
for char in name:
    if char.isupper():
        upper += 1
    elif char.islower():
        lower += 1
print("Uppercase :", upper)
print("Lowercase:", lower)
```

```python
num = int(input("enter the any number:"))
prime = True
if num < 2:
    prime = False
else:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            prime = False
            break
if prime:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```
