```python
import pandas as pd
```

```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
url="https://pharmeasy.in/health-care/sports-nutrition-12931"
header={
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/533.36"
}
response=requests.get(url,headers=header)
response.status_code==200
soup=BeautifulSoup(response.content,'html.parser')
parent_class="ProductCard_productCard__ftHIX"
print(parent_class)
products=soup.find_all("div",class_="ProductCard_productCard__ftHIX")
print(len(products))
df = pd.DataFrame()
name_list= []
price_list = []
for product in products:
    name=product.find("div",class_="ProductCard_productName__nWQ3x")
    price=product.find("div",class_="ProductCard_priceContainer__99QdN ProductCard_otcListingPriceContainer__PXAT_")
    if name!=None:
        name=name.get_text()
    if price!=None:
        price=price.get_text()

    if (price!=None) and (name!=None):
        price_list.append(price)
        name_list.append(name)
df["Name"]=name_list
df["Price"]=price_list
df
df.to_csv("Pharmeasy_data.csv", index=False)
```
